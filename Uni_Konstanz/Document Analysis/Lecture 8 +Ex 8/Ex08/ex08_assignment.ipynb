## Task 1: Naive Bayes Classifier (8 Pts)

There are four sentences in your training corpus:

1. The cat chases birds. ==> A
2. The cat sleeps. ==> A
3. The dog barks at the cat. => B
4. The cat steps on the sleeping dog. ==> B

Determine whether the sentence "The dog runs after the cat" is classified as "A" or "B" by a naive bayes classifier.

Apply any NLP pre-processing steps that you deem appropriate; e.g. stopword removal, stemming, etc.

Stopword remove, then we get:

1. cat chases birds. ==> A
2. cat sleeps. ==> A
3. dog barks cat. => B
4. cat steps sleeping dog. ==> B
5. dog runs cat ==> ?

Vocabulary = {cat, chases, birds, sleeps, dog, barks, steps, sleeping, runs}

Class A:

Words = {cat(2), chases(1), birds(1), sleeps(1)}
Total = 5

Class B:

Words = {dog(2), barks (1), cat(2), steps(1), sleeping (1)}
Total = 7

Vocabulary size = 9

Likelyhood class A:

p(cat | A) = (2 + 1)/ (5 + 9) = 3 / 14
p(dog | A) = (0 + 1) / (5 + 9) = 1 / 14
p(runs | A) = (0 + 1) / (5 + 9) = 1 / 14 

p(A) = 3/14 * 1/14 * 1/14 = 3 / (14^3)
Likelyhood class B:

P(cat | B) = (2 + 1) / (7 + 9) = 3/16
P(dog | B) = (2 + 1) / (7 + 9) = 3/16
P(runs| B) = (0 + 1) / (7 + 9) = 1 / 16
p(B) = 3 / 16 * 3/16 * 1/16 = 9 /( 16^3)

So p(B) > p (A) => B wins
