## Next Step: Assigning Zodiac Signs

The next stage of the research involves determining the zodiac sign for each individual in the dataset. Using the recorded dates of birth, each entry will be mapped to a corresponding astrological sign based on standard zodiac date ranges. This will produce an enhanced dataset with an additional categorical feature — **Zodiac Sign** — which will later be used for statistical analysis and comparison across professional domains.


```python
import csv 

def reading (path):

    dates = []

    with open(path, newline='') as file:
        csv_reader = csv.reader(file)
        for i in csv_reader:
            dates.append(i[-1])
    
    return dates[1::]



actors_dates = reading("data/actors_100_new.csv")
art_dates = reading("data/art_100_new.csv")
music_dates = reading("data/music_100_new.csv")
science_dates = reading("data/science_100_new.csv")
sports_dates = reading("data/sports_100_new.csv")


```

We now proceed to determine the zodiac sign for each individual in the dataset based on their date of birth.


```python
def signs (date_list):

    signs = []
    for i in date_list:
        
        i = i.split("-")
        month = i[1]
        day = int(i[2])
        
        if month == "01":
            if day >= 20:
                signs.append("Aquarius")
            else:
                signs.append("Capricorn")
        elif month == "02":
            if day >= 19:
                signs.append("Pisces")
            else:
                signs.append("Aquarius")
        elif month == "03":
            if day >= 21:
                signs.append("Aries")
            else:
                signs.append("Pisces")
        elif month == "04":
            if day >= 20:
                signs.append("Taurus")
            else:
                signs.append("Aries")
        elif month == "05":
            if day >= 21:
                signs.append("Gemini")
            else:
                signs.append("Taurus")
        elif month == "06":
            if day >= 21:
                signs.append("Cancer")
            else:
                signs.append("Gemini")
        elif month == "07":
            if day >= 23:
                signs.append("Leo")
            else:
                signs.append("Cancer")
        elif month == "08":
            if day >= 23:
                signs.append("Virgo")
            else:
                signs.append("Leo")
        elif month == "09":
            if day >= 23:
                signs.append("Libra")
            else:
                signs.append("Virgo")
        elif month == "10":
            if day >= 23:
                signs.append("Scorpio")
            else:
                signs.append("Libra")
        elif month == "11":
            if day >= 22:
                signs.append("Sagittarius")
            else:
                signs.append("Scorpio")
        elif month == "12":
            if day >= 22:
                signs.append("Capricorn")
            else:
                signs.append("Sagittarius")

    return signs
actors_signs = signs(actors_dates)
art_signs = signs(art_dates)
music_signs = signs(music_dates)
science_signs = signs(science_dates)
sports_signs = signs(sports_dates)
```

## Zodiac Sign Distribution Analysis

At this stage, we visualize the distribution of zodiac signs across the dataset.  
To achieve this, a **heatmap** will be generated based on the previously assigned zodiac information, allowing us to observe potential patterns or concentrations of specific signs among the studied individuals.


```python
zodiac_signs = ["Aries", "Taurus", "Gemini", "Cancer", "Leo", "Virgo", "Libra", "Scorpio", "Sagittarius", "Capricorn", "Aquarius", "Pisces"]

def count_popularity(signs_list):
    popularity = []
    for i in zodiac_signs:
        count = signs_list.count(i)
        popularity.append(count)

    return popularity


direction = ["actors", "art", "music", "science", "sports"]

popularity_of_sign_for_actors = count_popularity(actors_signs)
popularity_of_sign_for_art = count_popularity(art_signs)
popularity_of_sign_for_music = count_popularity(music_signs)
popularity_of_sign_for_science = count_popularity(science_signs)
popularity_of_sign_for_sports = count_popularity(sports_signs)

overall_popularity_of_sign_for_all = []

for i in range(0,12):
    overall_popularity_of_sign_for_all.append(popularity_of_sign_for_actors[i] + popularity_of_sign_for_art[i] + popularity_of_sign_for_music[i] + popularity_of_sign_for_science[i] + popularity_of_sign_for_sports[i])

```

```python
import numpy as np
import plotly.express as px


matrix = np.array(popularity_of_sign_for_actors).reshape(1, -1)
fig = px.imshow(
    matrix,
    labels=dict(x="Zodiac sign", y="", color="Popularity"),
    x=zodiac_signs,
    aspect="auto",
    color_continuous_scale="YlOrRd"
)

fig.update_layout(title=f"ASTROLOGY AND GENIUS FOR ACTORS")
fig.show()

matrix = np.array(popularity_of_sign_for_art).reshape(1, -1)
fig = px.imshow(
    matrix,
    labels=dict(x="Zodiac sign", y="", color="Popularity"),
    x=zodiac_signs,
    aspect="auto",
    color_continuous_scale="YlOrRd"
)

fig.update_layout(title=f"ASTROLOGY AND GENIUS FOR ART")
fig.show()

matrix = np.array(popularity_of_sign_for_music).reshape(1, -1)
fig = px.imshow(
    matrix,
    labels=dict(x="Zodiac sign", y="", color="Popularity"),
    x=zodiac_signs,
    aspect="auto",
    color_continuous_scale="YlOrRd"
)

fig.update_layout(title=f"ASTROLOGY AND GENIUS FOR MUSIC")
fig.show()

matrix = np.array(popularity_of_sign_for_science).reshape(1, -1)
fig = px.imshow(
    matrix,
    labels=dict(x="Zodiac sign", y="", color="Popularity"),
    x=zodiac_signs,
    aspect="auto",
    color_continuous_scale="YlOrRd"
)

fig.update_layout(title=f"ASTROLOGY AND GENIUS FOR SCIENCE")
fig.show()

matrix = np.array(popularity_of_sign_for_sports).reshape(1, -1)
fig = px.imshow(
    matrix,
    labels=dict(x="Zodiac sign", y="", color="Popularity"),
    x=zodiac_signs,
    aspect="auto",
    color_continuous_scale="YlOrRd"
)

fig.update_layout(title=f"ASTROLOGY AND GENIUS FOR SPORTS")
fig.show()

matrix = np.array(overall_popularity_of_sign_for_all).reshape(1, -1)
fig = px.imshow(
    matrix,
    labels=dict(x="Zodiac sign", y="", color="Popularity"),
    x=zodiac_signs,
    aspect="auto",
    color_continuous_scale="YlOrRd"
)

fig.update_layout(title=f"ASTROLOGY AND GENIUS FOR ALL")
fig.show()

```

### Summary Interpretation

Certain zodiac archetypes appear loosely reflected in the distribution results. For example:

- **Leo** shows a high concentration among actors, consistent with traits of charisma and public presence.
- **Pisces** and **Cancer** are prominent in creative fields, aligning with their emotional and artistic descriptions.
- **Capricorn** appears strongly in science-related domains, which aligns with themes of discipline and ambition.
- **Virgo**, traditionally associated with analytical strength, does not appear significantly in scientific fields within the dataset, contradicting the stereotype.

However, these patterns are **inconsistent across categories** and include unexpected deviations (e.g., Aquarius in sports). As such, they are more likely attributable to sample variation and chance rather than meaningful astrological influence.

### Conclusion

Although some superficial parallels between zodiac archetypes and domain frequency can be identified, the observed patterns are neither systematic nor statistically substantiated. Any perceived alignment appears coincidental and does not support the hypothesis that zodiac signs meaningfully correlate with fame or exceptional achievement across professional fields.
